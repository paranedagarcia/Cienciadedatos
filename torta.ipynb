```python
import matplotlib.pyplot as plt

labels = [
    "Relator",
    "Soporte integral",
    "Gestión de pagos",
    "Suscripciones de aplicaciones",
    "Servicios Cloud",
    "Difusión"
]

sizes = [65, 9, 4, 8, 10, 4]

plt.figure(figsize=(10, 8))
wedges, texts, autotexts = plt.pie(sizes, labels=labels, autopct='%1.0f%%')
for t in list(texts) + list(autotexts):
    t.set_fontsize(16)
plt.title("Distribución de Costos / Recursos", fontsize=16)
plt.show()
```
