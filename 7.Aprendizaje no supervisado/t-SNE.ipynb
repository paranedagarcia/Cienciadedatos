# t-SNE

El algoritmo **t-SNE** (del inglés *t-distributed Stochastic Neighbor Embedding* o Incrustaciones de Vecinos Estocásticos Distribuidos en t) es una técnica de **aprendizaje de variedades** (*manifold learning*) diseñada principalmente para la **visualización de datos de alta dimensionalidad** en espacios de dos o tres dimensiones. A diferencia de técnicas lineales como PCA, t-SNE se enfoca en preservar las distancias locales, intentando que los puntos que están cerca en el espacio original permanezcan cerca en la representación de baja dimensión.

A continuación, se presenta una descripción completa de sus características, funcionamiento y aplicaciones.

### Descripción del Algoritmo t-SNE

1.  **Naturaleza no lineal:** A diferencia del Análisis de Componentes Principales (PCA), que es una técnica lineal, t-SNE es un algoritmo **no lineal** capaz de capturar estructuras complejas en los datos que los métodos lineales suelen ignorar.

2.  **Objetivo principal:** Su propósito es encontrar una representación bidimensional o tridimensional de los datos que preserve, en la medida de lo posible, las **distancias locales** entre los puntos.

3.  **Preservación de vecinos:** La idea central es mantener cerca los puntos que están próximos en el espacio original de alta dimensión, mientras que los puntos alejados se mantienen alejados en la representación de baja dimensión. t-SNE pone mucho más énfasis en preservar la información de los **vecinos cercanos** que en mantener las distancias entre puntos que están muy separados.

4.  **Limitación técnica:** Una característica crítica de t-SNE es que **no permite transformar datos nuevos** (un conjunto de prueba); solo puede transformar los datos con los que fue entrenado originalmente. Por esta razón, se utiliza casi exclusivamente para el **análisis exploratorio de datos (EDA)** y no como un paso previo para el aprendizaje supervisado.

### ¿En qué casos se utiliza?

t-SNE se debe utilizar en las siguientes situaciones:
*   **Visualización de clusters:** Cuando se desea observar si existen agrupaciones naturales en un conjunto de datos complejo (como imágenes o texto) que no son evidentes en muchas dimensiones.

*   **Análisis Exploratorio de Datos:** Para obtener una "pista" visual sobre la estructura de los datos antes de aplicar otros algoritmos.

*   **Cuando PCA falla:** En casos donde las relaciones entre los datos son **no lineales** y el PCA (al ser lineal) mezcla las categorías en el gráfico, t-SNE suele ofrecer una separación mucho más clara.

### Ejemplos de Uso

*   **Reconocimiento de dígitos manuscritos (Dataset Digits):** Al aplicar PCA a este dataset, los números (del 0 al 9) suelen superponerse significativamente en el gráfico. Sin embargo, al utilizar **t-SNE**, el algoritmo logra separar las clases de forma casi perfecta en grupos densos, basándose únicamente en la similitud de los píxeles de las imágenes, sin conocer las etiquetas de antemano.

*   **Análisis de imágenes complejas:** Se emplea para visualizar cómo una red neuronal organiza internamente las características de imágenes de alta resolución, ayudando a entender qué puntos son considerados similares por el modelo.

*   **Bioinformática:** Para visualizar la estructura de datos genómicos o proteómicos, donde la dimensionalidad es extremadamente alta y las relaciones son inherentemente no lineales.

En resumen, mientras que PCA intenta rotar el dataset y mantener la varianza global, **t-SNE busca "desenrollar" la estructura interna** de los datos para que el ojo humano pueda distinguir grupos de interés en un plano de dos dimensiones.

Ejemplo completo y autocontenido utilizando el dataset Digits de scikit-learn, el cual es ideal para esta técnica ya que contiene imágenes de 8×8 píxeles de dígitos manuscritos (0-9) representados como vectores de 64 dimensiones.

```python
# 1. Importación de librerías necesarias [6, 7]
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE

# 2. Carga y preparación del dataset [4, 5]
# El dataset contiene 1,797 muestras de dígitos de 8x8 píxeles
digits = load_digits()
X = digits.data  # Características (intensidad de píxeles)
y = digits.target  # Etiquetas (el dígito real del 0 al 9)

# 3. Configuración y aplicación de t-SNE [8]
# t-SNE no tiene un método 'transform' por separado, por lo que usamos 'fit_transform'
# n_components=2 define que queremos una visualización bidimensional
tsne = TSNE(n_components=2, random_state=42)
digits_tsne = tsne.fit_transform(X)

# 4. Visualización de los resultados [9, 10]
plt.figure(figsize=(10, 8))
colors = ["#476A2A", "#7851B8", "#BD3430", "#4A2D4E", "#875525",
          "#A83683", "#4E655E", "#853541", "#3A3120", "#535D8E"]

# Representamos cada punto con su dígito correspondiente para mayor claridad
for i in range(len(X)):
    plt.text(digits_tsne[i, 0], digits_tsne[i, 1], str(y[i]),
             color=colors[y[i]],
             fontdict={'weight': 'bold', 'size': 9})

plt.xlabel("t-SNE característica 0")
plt.ylabel("t-SNE característica 1")
plt.title("Visualización de dígitos manuscritos usando t-SNE")
plt.show()
```

**Análisis del Resultado y Conceptos Clave**

• **Agrupación Natural**: Al observar el gráfico resultante, notarás que las clases (dígitos del 0 al 9) se separan en clústeres densos y bien definidos. Esto es notable porque t-SNE es un método no supervisado que no utilizó las etiquetas y para realizar el mapeo, sino solo la similitud intrínseca de los datos.

• **t-SNE vs. PCA**: Si realizaras este mismo ejercicio con PCA, verías que muchos dígitos se superponen significativamente, ya que PCA (al ser lineal) no puede capturar la estructura compleja y no lineal de estos datos tan bien como t-SNE.

• **Limitación de Transformación**: Es importante recordar que t-SNE no permite transformar datos nuevos que no estaban en el conjunto de entrenamiento original; su uso principal es la exploración de datos y la detección de estructuras ocultas.

• **Importancia del Escalado**: En aplicaciones reales con características de diferentes magnitudes, es fundamental aplicar estandarización (como StandardScaler) antes de t-SNE para evitar que una variable domine el cálculo de distancias.

**Analogía:** Imagina que intentas aplanar una hoja de papel arrugada que tiene puntos dibujados. PCA intentaría simplemente aplastar la hoja desde arriba (proyección lineal), lo que mezclaría puntos que estaban en pliegues diferentes. t-SNE, en cambio, intenta "desenrollar" con cuidado la hoja manteniendo cerca los puntos que realmente estaban dibujados uno al lado del otro, permitiéndote ver la estructura real del dibujo en 2D.
